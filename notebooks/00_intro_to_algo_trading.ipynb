## The idea in one paragraph

**Algorithmic trading** means writing down your trading decisions as exact rules a computer can follow — *"buy when X happens, sell when Y happens"* — instead of deciding by gut feeling. Because the rules are exact, you can test them on years of historical prices before risking any money. That test is called a **backtest**, and it's what this workshop is about.

## How a strategy goes from idea to market

1. **Idea** — a hunch about how prices behave, e.g. *"prices that are trending up tend to keep going up"*.
2. **Rules** — turn the hunch into something precise, e.g. *"hold the stock while its 50-day average is above its 200-day average"*.
3. **Backtest** — simulate those rules on historical data and measure the results honestly.
4. **Paper trading** — run the rules on live prices with fake money.
5. **Live trading** — real money, small at first, constantly monitored.

Most ideas die at step 3 — and that's the point. A good backtest saves you from losing money on a bad idea.

## Who does this?

- **Quant hedge funds and asset managers** — run systematic strategies at scale.
- **Banks** — use algorithms to execute large client orders and manage risk.
- **Market makers / trading firms** — continuously quote buy and sell prices and profit from the spread.
- **Individuals** — anyone with Python and data can backtest ideas, like you're about to.

## Key terms (one sentence each)

**Data**
- **Ticker** — the short code for something you can trade, e.g. `SPY` or `AAPL`.
- **ETF** — an exchange-traded fund: one ticker that holds a whole basket of assets (SPY holds the ~500 companies of the S&P 500).
- **OHLCV** — Open, High, Low, Close, Volume: the five numbers that describe one day of trading.
- **Adjusted price** — price history rewritten so stock splits and dividends don't show up as fake jumps.
- **Return** — the percentage change in value; going from 100 to 105 is a +5% return.

**Trading**
- **Long** — owning an asset, so you profit when it goes up.
- **Short** — borrowing an asset and selling it, so you profit when it goes down.
- **Position** — what you hold right now: long, short, or nothing (cash).
- **Signal** — the output of your rules telling you which position to hold.
- **Strategy** — the complete set of rules that turns data into positions.
- **Transaction costs** — fees and price spreads you pay every time you trade.
- **Basis point (bp)** — one hundredth of a percent (0.01%); 5 bps = 0.05%.

**Testing and measuring**
- **Backtest** — a simulation of how a strategy would have performed on historical data.
- **Benchmark** — what you compare against; here, simply buying and holding the asset.
- **Equity curve** — your portfolio's value over time.
- **Volatility** — how much returns bounce around; the most common measure of risk.
- **Sharpe ratio** — return earned per unit of volatility; higher is better.
- **Drawdown** — how far your portfolio has fallen below its previous peak.

**Classic mistakes**
- **Look-ahead bias** — accidentally using information your strategy couldn't have known at the time.
- **Overfitting** — tuning rules so closely to past data that they capture noise rather than a real pattern.
- **Survivorship bias** — testing only on assets that survived to today, ignoring the ones that failed.

**Indicators we'll use**
- **Moving average (MA)** — the average of the last N closing prices; it smooths out daily noise.
- **RSI** — Relative Strength Index, a 0–100 score of recent moves; below 30 is "oversold", above 70 "overbought".
